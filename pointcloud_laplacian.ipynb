# Denoise and Deform 3D Point Clouds in Python With Graph Laplacians

A **point cloud** is a set of data points in a 3D coordinate system, where each point represents a single spatial measurement on some object's surface, forming together the external surface of an object in space. 
Point cloud data can be generated by 3D laser scanners (LiDAR) or photogrammetry and are often used to create precise digital 3D representations used in construction, engineering, and manufacturing.

We often want to analyse local geometric properties of surfaces — how they bend, change slope, or vary in density. This is essential for creating realistic graphics, analyzing 3D scans, and optimizing shapes for engineering and the mathematical device for achieving this goal is computation of derivatives on such 3D objects. 

The key difference between point clouds and 3D meshes is that point clouds are unconnected data points, rather than a surface constructed from triangles or polygons (Figure 1).

Unlike mesh models, point clouds lack explicit connectivity information (edges and faces), making it difficult to compute derivatives on them [1]. A major challenge is to extract global information from the given point clouds due to the lack of connectivity. The graph Laplacian solves this by defining connectivity based on neighborhood graphs (e.g., $k$-nearest neighbors) and constructing a discrete operator that approximates continuous geometric properties, such as curvature or surface normal, on the discrete data. 

The applications of the graph Laplacian are numerous  - from LiDAR-based SLAM & navigation for umanned vehicles to surface reconstruction and mesh generation for hole-filling and repairing 3D scans.


In this tutorial we will compute one of the versios of the graph Laplacian and use it for smoothing and transforming one point cloud shape to another.

### What you will build  


The **graph Laplacian** is a differential operator defined on a graph constructed from points (as opposed to a mesh surface or continuous domain) that provides a discrete diffusion operator on a neighborhood graph, which we can use for smoothing and regularizing deformations. It is the core component behind smoothing, denoising, spectral shape processing, and "Laplacian editing"-style deformations. In this tutorial, we will compute it to transform some point clouds from the Stanford 3D Scanning Repository[3].

The task can be broke down into 6 steps:

1. Build a graph on the points uding the $k$ nearest neighbors (kNN).
2. Build a sparse matrix $W$ by computing weights $w_{ij}$ for neighbor pairs.
3. Compute the degree matrix $D$ by summing the wights row-wise.
4. Compute the Laplacian $L=D-W$ (unnormalized) or $L_{norm} = \mathbf{I} - D^{-1/2} W D^{-1/2}$ (normalized).
5. Deform shape 1 into shape 2 by minimizing the energy.  
6. Visualize the deformation process with Plotly.


### Point clouds and kNN graphs: from kNN to sparse adjacency 


Let $P = \{p_i\}_i^n$ with $p_i \in \mathbb{R}^3$ be a point cloud.  


For the computations and visuzlization we will be using several Python libraries:

```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as sl
from sklearn.neighbors import NearestNeighbors
import plotly.graph_objects as go
import open3d as o3d
from tests.tests import *
```

The last import is the `tests.py` file containing the unit tests.

Since a point cloud has no inherent connectivity, to compute a Laplacian, we need to "simulate a surface" by defining which points are neighbors. We will do so by constructing the $k$-Nearest Neighbor (k-NN) graph. For each point $P_i$ in the initial shape, find its $k$ nearest neighbors (e.g. $k=20$).

The neighbors define the local topology via the weight matrix $W \in \mathbb{R}^{n \times n}$ with elements
$$
w_{ij}=\begin{cases} 
 e^{-\frac{||p_i - p_j||^2}{2\sigma^2}} \quad \text{ if $i, j$ neighbors}\\ 
 0 \quad \quad \quad \quad \text{ otherwise.} \end{cases}
$$
Therefore, $W$ contains the adjacency information as well as the local structure information, where the closest neighbors are given more weight. Since we only search for $k \ll n$ closest points, $W$ will be sparse.

Another matrix we need to compute at this stage is a degree matrix $D$, which is a diagonal matrix, with nonzero elements $d_{ii} = \sum_j w_{ij}$ containing the sum of weights for point $i$.

Next, we compute the graph Laplacian matrix $L=D-W$ (the unnormalized Laplacian variant), which captures the local shape by comparing a point to the average of its neighbors. We will include the option of comparing the unnormalized variant to the normalized Laplacian $L_{norm} = \mathbf{I} - D^{-1/2} W D^{-1/2}$, whch can sometimes be more robust to irregular sampling.

### Graph Laplacian computation

For performing these steps in Python, we will define a function `build_knn_laplacian()`. It will take in the the point cloud data as a matrix, the number of neighbors, and an optional sigma value for the Gaussian weights computation, a $W$ symmetrization option, as well as the Laplacian normalization option. The function will build the graph by identifying the neighbors with `sklearn.neighbors.NearestNeighbors`, compute the Euclidean distances between the neighboring points and build the sparse matrix $W$ by applying the Gaussian function `weights = np.exp(-(dists**2) / (2.0 * sigma**2))` element-wise to the matrix of distances between the neighbors. Then, we'll define the diagonal matrix $D$ by summing the weights for each row of $W$ and compute two variants of Laplacian normalized and unnormalized.

```python
def build_knn_laplacian(P, k=20, sigma=None, symmetrize=True, normalize=False):
    
    n = P.shape[0]
    nn = NearestNeighbors(n_neighbors=k+1, algorithm="auto").fit(P)
    dists, inds = nn.kneighbors(P)

    # drop self neighbor at col 0
    dists = dists[:, 1:]
    inds = inds[:, 1:]

    # set default sigma if not provided (median neighbor distance)
    if sigma is None:
        sigma = np.median(dists)
        sigma = max(sigma, 1e-8)

    # weights: Gaussian of distance between neighbors
    weights = np.exp(-(dists**2) / (2.0 * sigma**2))

    # build sparse W
    rows = np.repeat(np.arange(n), k)
    cols = inds.reshape(-1)
    vals = weights.reshape(-1)

    W = sp.coo_matrix((vals, (rows, cols)), shape=(n, n)).tocsr()

    if symmetrize:
        # make undirected with max(W, W^T)
        W = W.maximum(W.T)

    d = np.array(W.sum(axis=1)).reshape(-1)
    d = np.maximum(d, 1e-12)
    D = sp.diags(d, offsets=0, shape=(n, n), format="csr")
    
    if normalize:
        D_inv_sqrt = sp.diags(1.0 / np.sqrt(d))
        L = sp.eye(n, format="csr") - (D_inv_sqrt @ W @ D_inv_sqrt)
    else:
        L = D - W
    
    return L, d
```

### Stable smoothing + simple nearest neighbor pull (explicit solve)

Now that we have computed the graph Laplacian, we need to understand how it acts on the data.

For a matrix of stacked point coordinates $P \in \mathbb{R}^{n \times 3}$, when $L$ acts on each coordinate dimension, for each point $P_i$ we obtain the Laplacian coordinates
$$
\delta_i =  \sum_j w_{ij} (p_i - p_j),
$$
containing the information about the "local detail" or curvature at each point. 


We can use this information to transform one point cloud shape, say $P \in \mathbb{R}^{n \times 3}$ to another, say $T \in \mathbb{R}^{m \times 3}$ . The key idea behind it is to
keep $\delta_P$ (local details of the initial point cloud shape) while gradually moving toward the target shape.
This is typically achieved by energy minimization for each time step $t = 1, \dots, \mathcal{T}$:
$$
\min_{P_t} \big( ||L P_t - \delta(t)||^2 + \lambda||P_t - T(t)||^2 \big), \quad \quad  (*)
$$
where $P_t$ are the point positions at time $t$, $\delta(t) = (1-t) \delta_P + t \delta_T$ interpolates Laplacian coordinates between the initial shape and the target, $T(t) = (1-t)P + t T_{corr}$ a positional interpolation target, 
$\lambda$ acts as a "stiffness" parameter: a high $\lambda$ makes the shape match the target perfectly but might warp the local details. A low $\lambda$ keeps the shape rigid (close to the initial shape).
The first summand represents the preservation of the local differentials (smoothing or denoising) while the second summand -- the pull towards the target positions

(*) is a standard least squares with regularization problem, which has a closed-form solution. The normal equations are:

$$
(L^T L +\lambda I)P' = L^T \delta + \lambda T.   \quad \quad  (**)
$$


So far we have not defined the correspondence matrix $T_{corr}$ and the interpolated Laplacian coordinates $\delta$. $T_{corr}$ establishes the correspondence reflecting which point on the initial shape $P$ wants to move toward which point on the target shape $T$. This can be done in many ways. If the clouds have the same ordering and size, one can use a 1-to-1 mapping. However, such situations are rare and nearest neighbor mapping from $P$ to $T$ can be used. Another option would be nonrigid registration (CPD, ICP variants). To keep things simple, we will go with kNN again. 


Computing $T_{corr}$ via nearest-neighbor projection, naturally supports different point counts by building a function
$$
\phi:\{1,\dots, n\} \rightarrow \{1,\dots,m\},
$$
where for each point $p_i$ we pick the closest target point: $\phi(i) = \text{argmin}_{j \in 1,\dots,m}||p_i - t_j||$ and define the corresponding target position as $t^{corr}_i = t_{\phi(i)}$, thus $T_{corr} \in \mathbb{R}^{n \times 3}$. So, in this setup, both many-to-one and one-to-many mappings work.

We will define the function for recomputing the correspondence using the kNN algorithm, which we will need later.

```python
def recompute_Tcorr(Pcur, T):
    nn = NearestNeighbors(n_neighbors=1, algorithm="auto").fit(T)
    _, idx = nn.kneighbors(Pcur)
    return T[idx[:, 0]]
```

The differential displacement at time $t$, $\delta(t)$, is a simple linear interpolation between the displacement vectors of the source ($\delta_P$) and the target ($\delta_T$) at each point in the cloud.

To implement this step in Python, we will define the `laplacian_morph_frames()` function that takes in the initial data matrix $P$ and the target $T$, stiffness parameter $\lambda$, the number of frames (time steps), the refresh interval for the Laplacian and the correspondence, and optionally the sigma for computing the Laplacian.
After computing the differential coordinates  `deltaP = L_current @ Pcur` and `deltaT = L_current @ Tcorr`, we will solve (**) by inverting the smoothing term preceeding $P'$ on the left-hand side.

For solving the normal equations we will be using the LU decomposition for the more convenient solution of the system of equations with `scipy.sparse.linalg.factorize()`, which works well on sparse matrices. Then we use the `solve` method, which uses the LU decompositionn for matrix inversion. 

Note that we include the Laplacian recomputation `build_knn_laplacian(Pcur, k=k, sigma=sigma, symmetrize=True)` every `refresh_every` steps into the timeframe loop: as we deform `cPcur`, we also update the neighborhood graph so that the “surface connectivity” we are smoothing over is not frozen in the initial geometry. If we didn't refresh, the connections between points would be based on the original shape, even as it deforms into the target shape, which wouldn't make physical sense. Refreshing allows the 'mesh' to adapt to the new shape."

```python
def laplacian_morph_frames(
    P, T, k=20, lam=10.0, n_frames=30,
    refresh_every=5, sigma=None, normalize=True
):

    n = P.shape[0]
    I = sp.eye(n, format="csr")

    ts = np.linspace(0.0, 1.0, n_frames)
    frames = []

    Pcur = P.copy()

    # Placeholders for updates
    L_current = None
    solve_P = None
    deltaP = None
    deltaT = None
    Tcorr = None

    for i, t in enumerate(ts):
        if (i % refresh_every == 0) or (solve_P is None):
            # 1) Refresh correspondences using current geometry
            Tcorr = recompute_Tcorr(Pcur, T)  # (n,3)

            # 2) Refresh Laplacian (graph) on current geometry
            L_current, _ = build_knn_laplacian(Pcur, k=k, sigma=sigma, symmetrize=True, normalize=normalize)

            # 3) Refresh Laplacian coords + solver factorization
            deltaP = L_current @ Pcur
            deltaT = L_current @ Tcorr

            A = (L_current.T @ L_current) + lam * I
            solve_P = sl.factorized(A.tocsc())

        # Interpolate *targets*
        delta_t = (1.0 - t) * deltaP + t * deltaT
        Tt = (1.0 - t) * P + t * Tcorr

        rhs = (L_current.T @ delta_t) + lam * Tt

        Pnext = np.empty_like(Pcur)
        for dim in range(3):
            Pnext[:, dim] = solve_P(rhs[:, dim])

        frames.append(Pnext)
        Pcur = Pnext

    return ts, frames
```

### Deformation visualization

For visualization of the transformation process, we will use Plotly 3D graph with a slider to loop over the deformation frames as time steps. We will be plotting the pointcloud by calling `graphical_objects.Scatter3d()` with x, y, and z coordinates of the points: initialize plotly `graphical_objects.Figure`, set the initial frame (`frames[0]`), define the slider by feeding the rest 1 through `n_frames` by appending them and defining slider steps for the animation, then updating layout to create the slider and the control buttons ("Pause", "Play")  and set the desired camera view. The color of each point is based on its distance from the origin. This helps track the motion of different parts of the shape.

```python
def plotly_slider(frames, title="Laplacian Morph", marker_size=2):

    # Stable camera bounds across all frames
    all_pts = np.vstack(frames)
    xmin, ymin, zmin = all_pts.min(axis=0)
    xmax, ymax, zmax = all_pts.max(axis=0)

    def dist_from_origin(X):
        return np.sqrt((X**2).sum(axis=1))

    # Initial frame 
    X0 = frames[0]
    c0 = dist_from_origin(X0)

    fig = go.Figure(
        data=[
            go.Scatter3d(
                x=X0[:, 0], y=X0[:, 1], z=X0[:, 2],
                mode="markers",
                marker=dict(
                    size=marker_size,
                    color=c0,
                    colorscale="Viridis",
                    showscale=True,
                    colorbar=dict(title="||x||")
                ),
                name="Deformed"
            )
        ]
    )

    # Animation frames (update positions)
    plotly_frames = []
    for i, Xt in enumerate(frames):
        ct = dist_from_origin(Xt)
        plotly_frames.append(
            go.Frame(
                data=[go.Scatter3d(
                    x=Xt[:, 0], y=Xt[:, 1], z=Xt[:, 2],
                    marker=dict(color=ct) 
                )],
                name=str(i)
            )
        )

    fig.frames = plotly_frames

    # Slider steps
    steps = [
        dict(
            method="animate",
            args=[[str(i)], dict(mode="immediate",
                                frame=dict(duration=0, redraw=True),
                                transition=dict(duration=0))],
            label=str(i),
        )
        for i in range(len(frames))
    ]

    fig.update_layout(
        title=title,
        sliders=[dict(active=0, currentvalue={"prefix": "frame: "}, pad={"t": 30}, steps=steps)],
        updatemenus=[dict(
            type="buttons",
            showactive=False,
            buttons=[
                dict(label="Play", method="animate",
                     args=[None, dict(frame=dict(duration=60, redraw=True),
                                      transition=dict(duration=0),
                                      fromcurrent=True)]),
                dict(label="Pause", method="animate",
                     args=[[None], dict(frame=dict(duration=0, redraw=False),
                                       mode="immediate")]),
            ],
            x=0.0, y=1.0
        )],
        scene=dict(                       # set camera view 
        xaxis=dict(autorange='reversed'), 
        camera=dict(
            up=dict(x=0, y=1, z=0),      
            eye=dict(x=0, y=0, z=2),      
            center=dict(x=0, y=0, z=0)
        ),
        dragmode='orbit'                
    )
    )
    
    fig.show()
```

Finally, we just create the wrapper for all the above functions that will take in the desired point cloud data (initial and target, set the number of neighbors for the kNN algorithm, the stiffness parameter controlling the morph process and the number of time steps), normalize the point clouds, find the correspondence between the initial points and the target points, compute the laplacian, solve the energy minimization and visualize the results. 

We will use the `o3d.io.read_point_cloud()` function to read in the point cloud data and then transform it into a `float64` numpy array. In many applications, before computing the graph Laplacian, it is beneficial to normalize the data to a unit sphere. The best practice in deformation applications is to normalize with the same transform, applying the center/scale of the initial shape to each cloud. Normalization is critical for computation of adjacency weights during the graph construction (which assumes roughly the same coordinate frame for the initial and target shapes), as well as for the target attraction force magnitude (the step size depends on the coordinate scale).

```python
# Load the data
pcd_init = o3d.io.read_point_cloud("data/bunny.ply")
initial_points = np.asarray(pcd_init.points, dtype=np.float64)

pcd_tgt = o3d.io.read_point_cloud("data/dragon.ply")
target_points = np.asarray(pcd_tgt.points, dtype=np.float64)


def morph_ply_pointclouds(P, T, k=20, lam=10.0, n_frames=11, 
                          refresh_every=5, sigma=None, normalize=True):

    # Normalize
    mu = P.mean(axis=0, keepdims=True)
    P0 = P - mu
    scale = np.sqrt((P0**2).sum(axis=1)).mean()

    P = P0 / (scale + 1e-12)
    T = (T - mu) / (scale + 1e-12)
    
    # Compute deformation frames
    ts, frames = laplacian_morph_frames(P, T, k=k, lam=lam, n_frames=n_frames,
    refresh_every=refresh_every, sigma=sigma, normalize=normalize)
    
    # Visualize
    plotly_slider(frames, title=f"Laplacian morph")
```

Now all we have to do is to call the `morph_ply_pointclouds` funciton and observe the deformation!  


```python
morph_ply_pointclouds(initial_points, target_points)
```

### Testing 

For a Laplacian point cloud morph, good unit tests are the ones that catch wrong sparsity pattern, broken symmetry, disconnected graphs, unstable solves, and correspondence mistakes.
We will go through 6 useful tests that will help to catch a few things that can silently go wrong.


1) Laplacian construction tests (Shape + sparsity + properties)

**Test 1** checks the Laplacian shape: should be $n \times n$.

**Test 2** checks sparsity: the umber of nonzeros is expected to be ~ $n \cdot k$ after symmetrization.

**Test 3a** checks the row-sum zero property: for the unnormalized graph Laplacian $L=D−W$ we expect:
$L\mathbf{I}=0$.

**Test 3b** checks the row-sum zero property: for the normalized graph Laplacian $L=\mathbf{I} - D^{-1/2}W D^{-1/2}$ we expect: $L_{norm}(D^{1/2} \mathbf{I})=0$.

2) Differential coordinates test

**Test 4** checks translation invariance invariance of the Laplacian coordinates. We expect
$L(X+c)=LX$, that is, that the coordinates will not change is the entire point cloud is translated by a scalar $c$. We test this by generating a random translation c and confirming that $L(X+c) ≈ LX$.

3) End-to-end behavioral tests (small synthetic clouds)

**Test 5** checks the “no-op” morph $P \rightarrow P$. We check that if we create a synthetic cloud (or take a given one), setting Set $T = P$, will produce every frame $\approx P$ (or extremely close).

**Test 6** checks a known deformation behavior. We create $T$ by scaling $z$ coordinate: $T[:,2] = 2 \cdot P[:,2]$. We expect that the distance $||P_t - T||$ monotonically decreases with $t$.


More useful verifications can be introduced (graph connectivity test, differential coordinates rotation test, correspondence tests, etc.).  

The above described 6 tests are gethered in `tests/tests.py`.
